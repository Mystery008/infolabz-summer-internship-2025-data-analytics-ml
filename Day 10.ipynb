### 🌲 Random Forest Classifier — In Brief

A Random Forest Classifier is a powerful and widely used machine learning algorithm for classification tasks. It belongs to the ensemble learning family, which means it combines multiple models to make better predictions.

### 🔍 How It Works:

1. Multiple Decision Trees:
It builds many decision trees (usually 100s) during training.

2. Randomness in Two Ways:
- Data Sampling: Each tree is trained on a random subset of the training data (bootstrapping).
- Feature Sampling: At each split in the tree, only a random subset of features is considered.

3. Voting:
For classification, each tree makes a prediction, and the forest picks the majority vote as the final output.

### ✅ Advantages:

- High accuracy in many cases.
- Handles non-linear relationships well.
- Robust to noise and overfitting, especially compared to a single decision tree.
- Works well with both categorical and numerical features.
- Provides feature importance scores.

### ⚠️ Disadvantages:
- Less interpretable than a single tree.
- Can be slower to train and predict, especially with many trees or large datasets.
- Large forests can consume more memory.



###  Implementation of a Random Forest Classifier using 'master_decision_file.csv' dataset 
- (GATE_Score and Salary to predict Should_Do_Masters):

### 1. Imports & Label Encoding

```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load
df1 = pd.read_csv('master_decision_file.csv')

# Encode categorical 
le = LabelEncoder()
df1['GATE_Score'] = le.fit_transform(df1['GATE_Score'])
df1['Should_Do_Masters'] = le.fit_transform(df1['Should_Do_Masters'])
```

### 2. Feature & Label Setup

```python
X = df1[['GATE_Score', 'Salary']]
Y = df1['Should_Do_Masters']
```

### 3. Train/Test Split

```python
X_train,X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=42)
```

### 4. Training Random Forest

```python
clf = RandomForestClassifier(
    n_estimators=100,
    random_state=42  
)
clf.fit(X_train, Y_train)
```

### 5. Model Evaluation

```python
prediction = clf.predict(X_test)
print(prediction)

accuracy = accuracy_score(Y_test, prediction)
accuracy
```

### 6. Feature Importance 

```python
import matplotlib.pyplot as plt
import pandas as pd

# Get feature importances from trained RandomForestClassifier
importances = pd.Series(clf.feature_importances_, index=X.columns)

# Plot as horizontal bar chart
importances.plot(kind='barh')
plt.title("Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

# Display raw importance values
clf.feature_importances_
```
